# 1970 lyrics filtering and tokenization

This notebook matches the streamed Genius lyrics CSV against `songs_1970.csv`, then tokenizes the matched lyrics with spaCy's `en_core_web_sm` model.

```python
%pip install -q spacy
!python -m spacy download en_core_web_sm

from filter_song_lyrics import filter_lyrics
from pathlib import Path

written = filter_lyrics(
    Path("data/song_lyrics.csv"),
    Path("songs_1970.csv"),
    Path("data/song_lyrics_1970.csv"),
)
print(f"Wrote {written} matching 1970 songs")
```

```python
from tokenize_1970_lyrics import tokenize
from pathlib import Path

written = tokenize(
    Path("data/song_lyrics_1970.csv"),
    Path("data/song_lyrics_1970_tokenized.csv"),
    "en_core_web_sm",
)
print(f"Tokenized {written} songs")
```

```python
import csv
import json
from collections import Counter


with open("data/song_lyrics_1970_tokenized.csv", encoding="utf-8", newline="") as f:
    row = next(csv.DictReader(f))

token_counts = Counter(json.loads(row["filtered_lyrics_tokens"]))

print(row["title"])
print(row["artist"])
print("Original token count:", row["token_count"])
print("Filtered token count:", row["filtered_token_count"])
print("Most common filtered tokens:")
for token, count in token_counts.most_common(20):
    print(f"{token!r}: {count}")
```

```python
dataset_word_counts = Counter()

with open("data/song_lyrics_1970_tokenized.csv", encoding="utf-8", newline="") as f:
    for row in csv.DictReader(f):
        dataset_word_counts.update(json.loads(row["filtered_lyrics_tokens"]))

print("Most common filtered words across the entire 1970 dataset:")
for word, count in dataset_word_counts.most_common(30):
    print(f"{word!r}: {count}")
```
